### 📌 Análisis individual [3.0 puntos]

#### 📌 Batch 1 (01/01/25 - 05/01/25) [0.75 puntos]

Este batch corresponde a la **primera semana de enero 2025 (W01)**. El modelo utilizado para generar estas predicciones fue entrenado con datos de las semanas W51 y W52 de diciembre 2024.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import json

# Configuración de paths
DATA_PATH = Path('data')
PRED_PATH = Path('predictions')
MLRUNS_PATH = Path('../airflow/mlruns/607081956794955500')

# Función para calcular métricas de evaluación
def evaluate_batch(data_file, pred_file, batch_name):
    """Evalúa un batch comparando predicciones con datos reales."""
    # Leer datos reales
    df_real = pd.read_parquet(DATA_PATH / data_file)
    real_pairs = set(zip(df_real['customer_id'], df_real['product_id']))
    
    # Leer predicciones
    df_pred = pd.read_csv(PRED_PATH / pred_file, header=None, names=['customer_id', 'product_id'])
    pred_pairs = set(zip(df_pred['customer_id'], df_pred['product_id']))
    
    # Calcular métricas
    tp = len(pred_pairs & real_pairs)
    fp = len(pred_pairs - real_pairs)
    fn = len(real_pairs - pred_pairs)
    
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0
    
    return {
        'batch': batch_name,
        'real_pairs': len(real_pairs),
        'pred_pairs': len(pred_pairs),
        'tp': tp, 'fp': fp, 'fn': fn,
        'precision': precision,
        'recall': recall,
        'f1': f1
    }

# Definir batches con sus archivos correspondientes según el enunciado
# Batch 1-4 corresponden a W01-W04 (enero 2025)
batches_info = {
    'Batch 1': {
        'data': 'batch_2025-W01.parquet',
        'pred': 'submission_2025-W01_predict_20251205_003438.csv',
        'run_id': '82869904c69941e1a68c5605e08ab4ed',
        'period': '01/01/25 - 05/01/25',
        'week': 'W01'
    },
    'Batch 2': {
        'data': 'batch_2025-W02.parquet', 
        'pred': 'submission_2025-W02_predict_20251205_003626.csv',
        'run_id': '3cf95a7c83ba449db0a27d4e7fd910fd',
        'period': '06/01/25 - 12/01/25',
        'week': 'W02'
    },
    'Batch 3': {
        'data': 'batch_2025-W03.parquet',
        'pred': 'submission_2025-W03_predict_20251205_003759.csv',
        'run_id': '752837ae4f40495aafaa40a629e33a47',
        'period': '13/01/25 - 19/01/25',
        'week': 'W03'
    },
    'Batch 4': {
        'data': 'batch_2025-W04.parquet',
        'pred': 'submission_2025-W04_predict_20251207_031232.csv',
        'run_id': 'c7cd6b955c0941569f0bdcead0ef716d',
        'period': '20/01/25 - 26/01/25',
        'week': 'W04'
    }
}

# ANÁLISIS BATCH 1 (W01) - Primera Semana de Enero
batch1 = batches_info['Batch 1']
metrics_b1 = evaluate_batch(batch1['data'], batch1['pred'], 'Batch 1')

print("BATCH 1: Primera Semana de Enero (01/01/25 - 05/01/25)")
print("-"*60)
print(f"\nMétricas de Predicción (comparación con datos reales):")
print(f"   Pares cliente-producto reales: {metrics_b1['real_pairs']:,}")
print(f"   Predicciones generadas: {metrics_b1['pred_pairs']:,}")
print(f"   Verdaderos Positivos (TP): {metrics_b1['tp']:,}")
print(f"   Falsos Positivos (FP): {metrics_b1['fp']:,}")
print(f"   Falsos Negativos (FN): {metrics_b1['fn']:,}")
print(f"\nMétricas de Evaluación:")
print(f"   Precision: {metrics_b1['precision']:.4f}")
print(f"   Recall: {metrics_b1['recall']:.4f}")
print(f"   F1-Score: {metrics_b1['f1']:.4f}")

# Cargar métricas del modelo desde MLflow
run_path = MLRUNS_PATH / batch1['run_id'] / 'artifacts' / 'metadata' / 'model_metadata.json'
with open(run_path) as f:
    model_meta = json.load(f)

print(f"\nMétricas del Modelo XGBoost (validación interna):")
for metric, value in model_meta['test_metrics'].items():
    print(f"   {metric}: {value:.4f}")

print(f"\nDecisión del Pipeline: {model_meta['comparison_result'].upper()}")

print(f"\nHiperparámetros Óptimos (Optuna - 50 trials):")
for param, value in model_meta['best_params'].items():
    if isinstance(value, float):
        print(f"   {param}: {value:.4f}")
    else:
        print(f"   {param}: {value}")

# Feature importance
fi_path = MLRUNS_PATH / batch1['run_id'] / 'artifacts' / 'metadata' / 'feature_importance.csv'
fi_df = pd.read_csv(fi_path)
print(f"\nTop 5 Features más importantes:")
for i, row in fi_df.head(5).iterrows():
    print(f"   {i+1}. {row['feature']}: {row['importance']:.4f}")
````

````python
# Visualización de Optuna y SHAP - Batch 1
from PIL import Image
import matplotlib.pyplot as plt

batch1_artifacts = MLRUNS_PATH / batch1['run_id'] / 'artifacts'

# Cargar importancia de hiperparámetros
with open(batch1_artifacts / 'optuna_plots' / 'hyperparameter_importance.json') as f:
    hp_importance_b1 = json.load(f)

print("Importancia de Hiperparámetros (Optuna fANOVA) - Batch 1:")
print("-"*50)
for hp, imp in zip(hp_importance_b1['hyperparameters'], hp_importance_b1['importances']):
    print(f"   {hp}: {imp:.4f} ({imp*100:.1f}%)")

# Mostrar gráficos SHAP en una fila
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

shap_files = ['shap_feature_importance.png', 'shap_summary.png', 'shap_waterfall.png']
shap_titles = ['SHAP Feature Importance', 'SHAP Summary Plot', 'SHAP Waterfall (Ejemplo)']

for ax, fname, title in zip(axes, shap_files, shap_titles):
    img = Image.open(batch1_artifacts / 'shap_plots' / fname)
    ax.imshow(img)
    ax.set_title(title, fontsize=11, fontweight='bold')
    ax.axis('off')

plt.suptitle('Interpretabilidad SHAP - Batch 1 (01/01/25 - 05/01/25)', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
````

### Análisis e Interpretación del Batch 1 (01/01/25 - 05/01/25)

El modelo generó 3,427 recomendaciones de pares cliente-producto para la primera semana de enero. Al contrastarlas con las 3,264 compras efectivas registradas, se identificaron 2,748 verdaderos positivos, 679 falsos positivos y 516 falsos negativos. Estos valores arrojan una cobertura (recall) de 0,8419, lo que significa que el sistema capturó el 84% de las compras reales. La precisión de 0,8019 indica que aproximadamente 8 de cada 10 recomendaciones emitidas correspondieron a productos que el cliente efectivamente adquirió.

El cociente entre predicciones generadas y compras reales es de 1,05 (3,427 / 3,264). Este valor indica que el modelo emitió prácticamente la misma cantidad de predicciones que compras reales hubo en el período. Un ratio cercano a 1 implica que el modelo es _conservador_: solo predice "compra" cuando tiene alta confianza, evitando saturar a los clientes con recomendaciones de baja probabilidad.

#### Análisis de Optuna (Optimización de Hiperparámetros)

El análisis fANOVA de Optuna revela que _gamma_ fue el hiperparámetro más determinante para el F1-Score en este batch, con una importancia del 54,8%. Gamma controla la regularización por complejidad del árbol: valores más altos penalizan la creación de nodos que no aportan ganancia suficiente. El valor óptimo encontrado (1,067) indica que el modelo necesitó regularización moderada para evitar sobreajuste a los patrones específicos de la primera semana post-festividades. El segundo hiperparámetro más relevante fue _colsample_bytree_ (21,8%), que controla qué fracción de features se usa en cada árbol.

#### Interpretación de Negocio basada en SHAP

Los valores SHAP revelan información crucial sobre el comportamiento de los clientes de SodAI Drinks en la primera semana de enero. La variable _brand_num_ se destaca como un factor influyente, lo que en el contexto B2B de distribución de bebidas refleja que los clientes mantienen acuerdos comerciales y preferencias de marca establecidas. Asimismo, el tamaño del producto (_size_) muestra una relación clara en el Summary Plot: los tamaños grandes aumentan la probabilidad de compra, sugiriendo que los clientes mayoristas reponen inventario en formatos grandes tras las festividades.

Adicionalmente, la ubicación geográfica (_Y_) influye significativamente, indicando que ciertas zonas concentran mayor actividad comercial en este periodo. Por último, la variable _customer_type_MAYORISTA_ aumenta la probabilidad de compra, consistente con el hecho de que los mayoristas poseen ciclos de reposición más predecibles.

#### 2. ¿En qué momento el modelo tuvo su peor desempeño y por qué?

El **Batch 2 (06/01/25 - 12/01/25)** tuvo el peor desempeño con F1-Score de 0.8041. Las razones se identifican en los artefactos:

*   **Evidencia del artefacto `feature_importance.csv`:** `sub_category_GASEOSAS` cayó de 25.9% a 8.5% de importancia, mientras que `package_KEG` subió de 16.3% a 19.3% (retorno del canal HORECA).

*   **Evidencia del SHAP Waterfall (`shap_waterfall.png`):** El ejemplo del Batch 2 muestra f(x)=2.273 con `size` dominando (+1.08), reflejando que el modelo priorizó correctamente la reposición de inventario en formatos grandes, pero la diversificación de demanda generó más falsos positivos.

---

#### 3. ¿Detectaron algún cambio significativo (drift) en la distribución de los datos? ¿Cómo lo identificaron?

Sí, el sistema detectó drift en _todos los batches_. El artefacto `model_metadata.json` registra `"comparison_result": "DRIFT_UPDATE"` en cada run.

El pipeline utiliza el test de Kolmogorov-Smirnov para comparar distribuciones de features entre el batch de referencia y el nuevo batch. Cuando el p-value es menor al umbral configurado, se activa el flag de drift. Los gráficos de _SHAP Summary Plot_ (`shap_summary.png`) muestran cambios en la distribución de valores SHAP entre batches, y el _Feature Importance_ (`feature_importance.csv`) muestra rotación de variables dominantes, como el paso de `sub_category_GASEOSAS` en Batch 1 a `size` en los Batches 3 y 4.

---

#### 4. ¿Tuvieron que reentrenar su modelo con los nuevos datos? ¿Por qué? ¿Ayudó esto al performance de su modelo?

_Sí_, el modelo se reentrenó automáticamente en todos los batches debido a la detección de drift. El artefacto `model_metadata.json` registra `"comparison_result": "DRIFT_UPDATE"`, indicando que el test estadístico detectó cambios significativos en la distribución de datos.

El reentrenamiento ayudó al performance, como evidencia la evolución de métricas: sin él, el F1-Score probablemente habría seguido cayendo después del Batch 2. Con reentrenamiento, el F1 mejoró de 0.8041 (Batch 2) a 0.8225 (Batch 4). Además, Optuna re-optimizó hiperparámetros en cada batch, adaptándose a los cambios, pasando de `gamma` dominante en Batch 1 a `learning_rate` en Batch 4.

---

#### 5. ¿Qué decisión técnica (modelo, métrica, imputación, etc.) tuvo más impacto en los resultados?

La _optimización automática de hiperparámetros con Optuna_ tuvo el mayor impacto. El análisis fANOVA muestra que diferentes hiperparámetros fueron críticos en cada batch. En Batch 3-4, `learning_rate` bajó significativamente, correlacionando con mejor F1-Score. Los gráficos de historia de optimización muestran convergencia en ~20-30 trials, mejorando consistentemente el F1 de validación.

La segunda decisión más importante fue usar _F1-Score como métrica de optimización_ en lugar de accuracy, dado el desbalance de clases inherente al problema de recomendación.

#### 6. ¿Qué hiperparámetro fue el más importante para su modelo?
