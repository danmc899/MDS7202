### **Producto de dataframes y creación de variable objetivo**

La idea es obtener una variable que nos diga si un cliente compró un producto en una semana. Para eso, juntamos los tres índices en una variable que diga `purchase_week`.

```python
dftransacciones['purchase_week'] = dftransacciones['purchase_date'].dt.to_period('W').apply(lambda r: r.start_time).dt.date
dftransacciones[['customer_id', 'product_id', 'purchase_date', 'purchase_week']]
```

```python
# Rango de semanas en el histórico
all_weeks = pd.period_range(
    start=dftransacciones['purchase_week'].min(),
    end=dftransacciones['purchase_week'].max(),
    freq='W'
).to_timestamp()

# Producto cartesiano cliente–producto–semana (variable objetivo)
clientes_productos = (
    pd.MultiIndex.from_product(
        [dfclientes['customer_id'].unique(),
         dfproductos['product_id'].unique(),
         all_weeks],
        names=['customer_id','product_id','purchase_week']
    ).to_frame(index=False)
)
clientes_productos
```

Acá se propone eliminar muchos ceros de productos que no fueron comprados por el cliente durante todas las semanas. La primera razón es porque no disponemos de la RAM para operar una tabla de 80 millones de registros. La segunda razón es porque no queremos eliminar series temporales que nos permitirían entender que algunos productos son más preferidos en cierta temporada (refrescos en verano, cafés en invierno). La tercera razón es porque de por sí existe un desbalanceo importante de datos, en donde hay productos que los clientes nunca van a comprar, dada sus preferencias individuales.

De esta forma, no castigamos a los modelos que entrenaremos después, que probablemente prefiera escoger la clase desbalanceada siempre para obtener resultados perfectos.

```python
import polars as pl
# ────────────────────────────────────────────────────────────────────────────
# 1. PREPARAR CASOS POSITIVOS
# ────────────────────────────────────────────────────────────────────────────

# Convertir transacciones a Polars LazyFrame para procesamiento eficiente
transacciones_lazy = pl.from_pandas(dftransacciones).lazy()

# Crear columna de semana de compra y obtener combinaciones únicas (cliente, producto, semana)
# que representan las compras reales (casos positivos)
compras_positivas_lazy = (
    transacciones_lazy
    .with_columns(
        pl.col("purchase_date").dt.truncate("1w").alias("purchase_week")
    )
    .select(["customer_id", "product_id", "purchase_week"])
    .unique()
    .with_columns(
        pl.lit(1).cast(pl.Int8).alias("compra_semanal")
    )
)

# Materializar los positivos para contar cuántos casos tenemos
compras_positivas = compras_positivas_lazy.collect()
n_positivos = len(compras_positivas)

# ────────────────────────────────────────────────────────────────────────────
# 2. GENERAR Y MUESTREAR CASOS NEGATIVOS (no-compras)
# ────────────────────────────────────────────────────────────────────────────

# Definir ratio: por cada compra real, tomamos 10 no-compras
ratio_negativos = 10
n_negativos = n_positivos * ratio_negativos

print(f"Compras reales (positivos): {n_positivos:,}")
print(f"Muestreando {n_negativos:,} no-compras (negativos)...")

# Crear el universo completo de combinaciones posibles (cliente × producto × semana)
df_universo_lazy = (
    pl.from_pandas(dfclientes).lazy()
    .select("customer_id").unique()
    .join(
        pl.from_pandas(dfproductos).lazy().select("product_id").unique(),
        how="cross"
    )
    .join(
        compras_positivas.lazy().select("purchase_week").unique(),
        how="cross"
    )
)

# Anti-join: quedarse solo con las combinaciones que NO fueron compradas
casos_negativos_lazy = df_universo_lazy.join(
    compras_positivas.lazy(),
    on=["customer_id", "product_id", "purchase_week"],
    how="anti"
)

# Muestrear aleatoriamente el número deseado de casos negativos
negativos_muestreados = (
    casos_negativos_lazy
    .select(pl.all().shuffle(seed=123).head(n_negativos))
    .with_columns(
        pl.lit(0).cast(pl.Int8).alias("compra_semanal")
    )
    .collect()
)

# ────────────────────────────────────────────────────────────────────────────
# 3. COMBINAR POSITIVOS Y NEGATIVOS, CONVERTIR A PANDAS
# ────────────────────────────────────────────────────────────────────────────

# Concatenar ambos conjuntos en Polars y convertir a Pandas para compatibilidad con scikit-learn
target_df = pl.concat([
    compras_positivas,
    negativos_muestreados
]).to_pandas()

# ────────────────────────────────────────────────────────────────────────────
# RESULTADO FINAL
# ────────────────────────────────────────────────────────────────────────────

print(f"\nDataFrame final: {len(target_df):,} filas")
print("\nDistribución de la variable objetivo:")
print(target_df["compra_semanal"].value_counts().sort_index())
```

Como se mencionó anteriormente, si no tenemos la ubicación de los individuos, no podemos aprovechar X ni Y para hacer análisis geoespacial. Por tanto, se propone eliminar estos clientes.

```python
# Unir target_df con información de clientes, productos y transacciones para obtener todas las features

# 1. Unir con dfclientes para agregar características del cliente
df_target = target_df.merge(
    dfclientes,
    on='customer_id',
    how='left'
)

# 2. Unir con dfproductos para agregar características del producto
df_target = df_target.merge(
    dfproductos,
    on='product_id',
    how='left'
)

# 3. Convertir purchase_week en dftransacciones a datetime para que coincida con target_df
dftransacciones_temp = dftransacciones.copy()
dftransacciones_temp['purchase_week'] = pd.to_datetime(dftransacciones_temp['purchase_week'])

# 4. Unir con dftransacciones para agregar items de las compras reales
# Solo los casos positivos (compra_semanal=1) tendrán estos valores,
# los negativos quedarán con 0
df_target = df_target.merge(
    dftransacciones_temp[['customer_id', 'product_id', 'purchase_week', 'items']],
    on=['customer_id', 'product_id', 'purchase_week'],
    how='left'
)

# Reemplazar NaN en 'items' por 0 para los casos negativos
df_target['items'] = df_target['items'].fillna(0)

print(f"\nDataFrame final con todas las features: {df_target.shape}")
print(f"\nPrimeras filas:")
display(df_target.head())

# Verificar valores nulos después del merge
print(f"\nValores nulos por columna:")
print(df_target.isnull().sum())
```

Como se mencionó anteriormente, si no tenemos la ubicación de los individuos, no podemos aprovechar X ni Y para hacer análisis geoespacial. Por tanto, se propone eliminar estos clientes.

En esta sección, se deben construir pipelines para automatizar el preprocesamiento de los datos, lo cual garantizará que el flujo de trabajo sea reproducible y eficiente para esta entrega y las futuras. El objetivo es aplicar una serie de transformaciones en un orden definido para asegurar que los datos estén listos para los modelos a entrenar. El pipeline final debe incluir las técnicas de pre-procesamiento que se deben aplicar a los distintos datos (según lo que consideren necesario para el problema). Por ejemplo:

- **Imputación de valores nulos**: Manejo de datos faltantes mediante estrategias adecuadas (media, mediana, moda, interpolación, etc.).

- **Transformaciones personalizadas**: Uso de ColumnTransformer para aplicar diferentes transformaciones a columnas específicas.

- **Codificación de variables categóricas**: Convertir datos categóricos a un formato numérico adecuado (One-Hot Encoding, Label Encoding, etc.).

- **Discretización de variables**: Conversión de variables numéricas continuas en categorías si son relevantes para el desempeño del modelo a entrenar.

- **Nuevas características** : Creación de variables adicionales que puedan aportar información relevante al modelo.

Cada una de estas transformaciones debe ser justificada en función de su relevancia para el problema y los datos, y es importante evaluar su impacto en el rendimiento del modelo. Además, el pipeline debe ser flexible y modular para poder probar diferentes configuraciones de preprocesamiento.

```python
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.preprocessing import StandardScaler

class CategoryEncoder(BaseEstimator, TransformerMixin):
    """
    Un transformador personalizado para codificar variables categóricas.
    - Aplica OrdinalEncoder a las columnas con una jerarquía definida.
    - Aplica OneHotEncoder a las columnas categóricas nominales.
    """
    def __init__(self, ordinal_cols=None, one_hot_cols=None):
        self.ordinal_cols = ordinal_cols if ordinal_cols is not None else []
        self.one_hot_cols = one_hot_cols if one_hot_cols is not None else []
        self.ordinal_encoder = None
        self.one_hot_encoder = None
        self.ohe_feature_names = []

    def fit(self, X, y=None, ordinal_categories=None):
        """
        Ajusta los codificadores a los datos.
        """
        # Configurar y ajustar OrdinalEncoder si se proporcionan las columnas y categorías
        if self.ordinal_cols and ordinal_categories:
            self.ordinal_encoder = OrdinalEncoder(categories=ordinal_categories)
            self.ordinal_encoder.fit(X[self.ordinal_cols])

        # Configurar y ajustar OneHotEncoder si se proporcionan las columnas
        if self.one_hot_cols:
            self.one_hot_encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
            self.one_hot_encoder.fit(X[self.one_hot_cols])
            self.ohe_feature_names = self.one_hot_encoder.get_feature_names_out(self.one_hot_cols)

        return self

    def transform(self, X):
        """
        Aplica las transformaciones de codificación a los datos.
        """
        X_transformed = X.copy()

        # Aplicar OrdinalEncoder
        if self.ordinal_encoder and self.ordinal_cols:
            X_transformed[self.ordinal_cols] = self.ordinal_encoder.transform(X_transformed[self.ordinal_cols])

        # Aplicar OneHotEncoder
        if self.one_hot_encoder and self.one_hot_cols:
            encoded_data = self.one_hot_encoder.transform(X_transformed[self.one_hot_cols])
            encoded_df = pd.DataFrame(encoded_data, columns=self.ohe_feature_names, index=X_transformed.index)

            # Eliminar columnas originales y concatenar las nuevas
            X_transformed = X_transformed.drop(columns=self.one_hot_cols)
            X_transformed = pd.concat([X_transformed, encoded_df], axis=1)

        return X_transformed

    def set_output(self,transform='default'):
      #No modificar esta función
      return self

class IQR(BaseEstimator, TransformerMixin):
    def __init__(self, columns = None, lam = 1.5):
        self.columns = columns
        self.lam = lam

    def fit(self, X, y=None):
        X = pd.DataFrame(X).copy()
        if self.columns is None:
            self.columns_ = X.select_dtypes(include=[np.number]).columns.tolist()
        else:
            self.columns_ = list(self.columns)

        # Calcular Q1, Q3 y IQR utilizando train
        q1 = X[self.columns_].quantile(0.25)
        q3 = X[self.columns_].quantile(0.75)
        iqr = q3 - q1
        self.lower_ = (q1 - self.lam * iqr).to_dict()
        self.upper_ = (q3 + self.lam * iqr).to_dict()
        return self


    def transform(self, X):
        X_transformed = pd.DataFrame(X).copy()
        for c in self.columns_:
            lower = self.lower_[c]
            upper = self.upper_[c]
            X_transformed[c] = X_transformed[c].clip(lower, upper)
        return X_transformed

    def get_feature_names_out(self, input_features=None):
        """Devuelve los nombres de las características de salida."""
        if input_features is None:
            return self.columns_
        return input_features

    def set_output(self,transform='default'):
    #No modificar esta función
        return self

class StandardScalerTransformer(BaseEstimator, TransformerMixin):
    """
    Un transformador personalizado para aplicar estandarización (StandardScaler)
    a las columnas numéricas.
    """
    def __init__(self):
        self.scaler = StandardScaler()
        self.columns = None

    def fit(self, X, y=None):
        # Guardar los nombres de las columnas para el transform
        self.columns = X.columns
        self.scaler.fit(X)
        return self

    def transform(self, X):
        # Aplicar la transformación y devolver un DataFrame
        X_scaled = self.scaler.transform(X)
        return pd.DataFrame(X_scaled, columns=self.columns, index=X.index)

    def set_output(self,transform='default'):
      #No modificar esta función
      return self

class MinMax(BaseEstimator, TransformerMixin):

    def fit(self,X):
        # Aseguro que es un dataframe
        X = pd.DataFrame(X).copy()
        # Guardo mínimos y máximos
        self.min_ = X.min()
        self.max_ = X.max()
        self.columns_ = X.columns
        return self

    def transform(self,X):
        # Conversión a dataframe
        X = pd.DataFrame(X).copy()
        # Fórmula del enunciado
        X_scaled = (X - self.min_) / (self.max_ - self.min_)
        return X_scaled

    def get_feature_names_out(self, input_features=None):
        """Devuelve los nombres de las características de salida."""
        return self.columns_

    def set_output(self,transform='default'):
        #No modificar este método
        return self

# variables dummies a int8
df_target['devolucion'] = df_target['devolucion'].astype('int8')
df_target['compra'] = df_target['compra'].astype('int8')
df_target['sin_transaccion'] = df_target['sin_transaccion'].astype('int8')

# Ver los tipos de datos de cada columna
print("Tipos de datos de df_target:")
print(df_target.dtypes)
```

A continuación, se detallan los hiperparámetros de XGBoost que se optimizarán:

- `max_depth`: Profundidad máxima de los árboles. Se optimiza para encontrar un balance entre la capacidad del modelo para capturar relaciones complejas y el riesgo de sobreajuste.
- `learning_rate`: Tasa de aprendizaje. Un valor más bajo requiere más árboles pero puede llevar a una mejor generalización. Se busca el valor óptimo que permita converger a una buena solución sin ser demasiado lento.
- `subsample`: Proporción de las muestras de entrenamiento utilizadas para construir cada árbol. Ayuda a prevenir el sobreajuste al introducir aleatoriedad.
- `colsample_bytree`: Proporción de las características (columnas) utilizadas para construir cada árbol. También es una técnica de regularización para evitar que el modelo dependa demasiado de unas pocas características.
- `min_child_weight`: Suma mínima de pesos de instancia necesarios en un nodo hijo. Un valor más alto hace que el algoritmo sea más conservador y evita que el modelo aprenda patrones muy específicos de los datos de entrenamiento.
- `gamma`: Reducción mínima de la pérdida requerida para hacer una partición. Actúa como un umbral de regularización; valores más altos conducen a menos divisiones y árboles más simples.
- `n_estimators`: Número de árboles (rondas de boosting). Se optimiza junto con `learning_rate` para encontrar la combinación que mejor rendimiento ofrezca sin sobreajustar ni tardar demasiado en entrenar.

La métrica que se busca optimizar es precision, porque se quiere evitar lo más posible la presencia de falsos positivos en el modelo, es decir se quiere evitar que tripletes cliente-producto-semana se clasifiquen como compra erróneamente ya que de lo contrario puede llevar a acumulación de productos sin vender, pérdida de la inversión inicial y vencimiento de los productos. Por esto se selecciona el modelo de ensamble `XGBoost`, el cual entregó mejores resultados en este ámbito.

```python
import numpy as np
import optuna
from sklearn.metrics import precision_score
from xgboost import XGBClassifier
from optuna.samplers import TPESampler

# f.o. para optuna
def objective_function(trial):
    params = {
        "objective": "binary:logistic",
        "eval_metric": "logloss",
        "max_depth": trial.suggest_int("max_depth", 3, 8),
        "learning_rate": trial.suggest_loguniform("learning_rate", 0.001, 0.1),
        "subsample": trial.suggest_float("subsample", 0.5, 1.0),
        "colsample_bytree": trial.suggest_float("colsample_bytree", 0.5, 1.0),
        "min_child_weight": trial.suggest_int("min_child_weight", 1, 7),
        "gamma": trial.suggest_float("gamma", 0, 1),
        "n_estimators": trial.suggest_int("n_estimators", 50, 300),
        "random_state": 42,
        "use_label_encoder": False
    }

    # Preprocesar los datos
    X_train_transformed = preprocessor.fit_transform(X_train)
    X_val_transformed = preprocessor.transform(X_val)

    # Entrenar con los datos ya preprocesados
    model = XGBClassifier(**params)
    model.fit(X_train_transformed, y_train)

    # Validar con los datos de validación ya preprocesados
    y_pred = model.predict(X_val_transformed)
    precision = precision_score(y_val, y_pred, pos_label=1)

    return precision

# Crear un muestreador con una semilla para obtener resultados consistentes
sampler = TPESampler(seed=42)

# Crear el estudio de Optuna, pasando el muestreador
study = optuna.create_study(direction="maximize", sampler=sampler)
study.optimize(objective_function, n_trials=25)

print("Mejor precisión:", study.best_value)
print("Mejores hiperparámetros:", study.best_params)
```

Se observa que se obtuvo una precision de 86%, que significa un incremento de 21% en comparación al entrenamiento sin seleccionar hiperparámetros en el modelo anterior. Para lograrlo, se entrenó el modelo con los siguientes hiperparámetros:
`{'max_depth': 8, 'learning_rate': 0.015930522616241012, 'subsample': 0.8540362888980227, 'colsample_bytree': 0.5102922471479012, 'min_child_weight': 7, 'gamma': 0.8324426408004217, 'n_estimators': 103}`

Se utiliza el resultado de la optimización con Optuna para construir y evaluar el modelo XGBoost final.

Los pasos son los siguientes:
1.  **Extracción de Hiperparámetros**: Se obtienen los mejores hiperparámetros (`best_params_from_optuna`) encontrados por el estudio de Optuna.
2.  **Combinación de Parámetros**: Se combinan los hiperparámetros optimizados con los parámetros fijos (como `random_state`) para crear la configuración final del modelo (`final_params`).
3.  **Creación del Pipeline Optimizado**: Se instancia un nuevo pipeline (`pipeline_xgboost_optimized`) que incluye el preprocesador y el clasificador `XGBClassifier` configurado con los parámetros óptimos.
4.  **Entrenamiento Final**: El pipeline optimizado se entrena con el conjunto de datos de entrenamiento (`X_train`, `y_train`).
5.  **Evaluación Exhaustiva**: El rendimiento del modelo final se evalúa tanto en el conjunto de validación como en el de prueba. Se imprimen el `classification_report` y la matriz de confusión para ambos conjuntos, lo que permite verificar la capacidad de generalización del modelo en datos no vistos.

```python
# 1. Extraer los mejores hiperparámetros del estudio de Optuna
best_params_from_optuna = study.best_params

# 2. Definir parámetros fijos que no estaban en la búsqueda de Optuna
fixed_params = {
    'random_state': 42,
    'use_label_encoder': False,
    'eval_metric': "logloss"
}

# 3. Combinar ambos diccionarios para tener la configuración completa del modelo
final_params = {**best_params_from_optuna, **fixed_params}

# 4. Crear un nuevo pipeline con el clasificador XGBoost optimizado
pipeline_xgboost_optimized = Pipeline([
    ('preprocessor', preprocessor),
    ("classifier", XGBClassifier(**final_params))
])

# 5. Entrenar el pipeline optimizado con los datos de entrenamiento
pipeline_xgboost_optimized.fit(X_train, y_train)
print("\nEntrenamiento finalizado.")

# 6. Evaluación en el conjunto de validación
print("Evaluación en validación para XGBoost optimizado")

y_val_pred_final = pipeline_xgboost_optimized.predict(X_val)

# Reporte de clasificación
print("\nReporte de Clasificación (Validación):")
print(classification_report(y_val, y_val_pred_final, target_names=["No Compra", "Compra"], digits=4))

# Matriz de confusión
tn_val, fp_val, fn_val, tp_val = confusion_matrix(y_val, y_val_pred_final).ravel()
print("Matriz de Confusión (Validación):")
print(f"TN: {tn_val:,} | FP: {fp_val:,} | FN: {fn_val:,} | TP: {tp_val:,}")


# 7. Evaluación en el conjunto de prueba
print("Evaluación en prueba para XGBoost optimizado")
y_test_pred_final = pipeline_xgboost_optimized.predict(X_test)

# Reporte de clasificación
print("\nReporte de Clasificación (Prueba):")
print(classification_report(y_test, y_test_pred_final, target_names=["No Compra", "Compra"], digits=4))

# Matriz de confusión
tn_test, fp_test, fn_test, tp_test = confusion_matrix(y_test, y_test_pred_final).ravel()
print("Matriz de Confusión (Prueba):")
print(f"TN: {tn_test:,} | FP: {fp_test:,} | FN: {fn_test:,} | TP: {tp_test:,}")
```

Para explicar el funcionamiento del modelo, se utilizaron gráficos de Waterfall para tres predicciones específicas. Estos gráficos descomponen la predicción del modelo en contribuciones individuales de cada atributo, permitiendo entender cómo cada uno influye en la decisión final. En específico: 

- **Valor Base E[f(x)] (Abajo)**: Es la predicción promedio del modelo para todas las muestras. Es el punto de partida. Un valor negativo indica que, en promedio, el modelo tiende a predecir "No Compra".
- **Barras Rojas (↑)**: Son los atributos que aumentan la probabilidad de compra para esa instancia específica. Empujan la predicción hacia un valor más positivo. El número junto a la barra es el valor SHAP de ese atributo.
- **Barras Azules (↓)**: Son los atributos que disminuyen la probabilidad de compra. Empujan la predicción hacia un valor más negativo.
- **Valor Final f(x) (Arriba)**: Es la predicción final del modelo para esa instancia, después de sumar todas las contribuciones (rojas y azules) al valor base. Un valor final positivo significa que el modelo predice "Compra" (1), y uno negativo, "No Compra" (0).

En este caso, se puede ver que en las primeras tres predicciones, la suma de las contribuciones de los atributos (barras rojas y azules) lleva a un valor final negativo, lo que significa que el modelo predice "No Compra" para estas instancias.
